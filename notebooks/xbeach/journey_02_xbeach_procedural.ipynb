# Journey 2 — Build an XBeach workspace procedurally

**Learning goals:** Create the time range and computational grid used by an XBeach configuration.

**Prerequisites:** Complete Journey 1 and have `rompy-xbeach` installed.

**Execution contract:** This lesson prepares configuration objects. It does not invoke the XBeach binary.

```python
from rompy.core.time import TimeRange
from rompy_xbeach.grid import GeoPoint, RegularGrid

times = TimeRange(start='2023-01-01T00:00', end='2023-01-01T06:00', interval='1h')
grid = RegularGrid(
    ori=GeoPoint(x=115.594239, y=-32.641104, crs='EPSG:4326'),
    alfa=347.0, dx=10.0, dy=15.0, nx=12, ny=10, crs='EPSG:28350',
)
print(times.duration)
print(f'{grid.nx} x {grid.ny} cells')
```
