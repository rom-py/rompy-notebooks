## Visual verification: bathymetry and grid

XBeach is sensitive to both the shape of the bathymetry and the resolution/orientation of the computational grid. A quick plot catches sign, extent, and shoreline mistakes before a model run.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 110, 12)
y = np.linspace(0, 90, 10)
xx, yy = np.meshgrid(x, y)
bathy = -0.5 - 0.08 * xx - 0.015 * (yy - 45) ** 2 / 10
fig, ax = plt.subplots(figsize=(8, 4))
mesh = ax.pcolormesh(xx, yy, bathy, shading="auto", cmap="terrain")
ax.contour(xx, yy, bathy, levels=[0], colors="red", linewidths=2)
ax.set(title="Teaching bathymetry used by the XBeach grid", xlabel="cross-shore distance (m)", ylabel="alongshore distance (m)")
fig.colorbar(mesh, ax=ax, label="elevation (m)")
plt.show()
```
