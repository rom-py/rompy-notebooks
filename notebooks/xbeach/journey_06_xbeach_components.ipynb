# Journey 6 — Configure XBeach physics, sediment, outputs, and hotstarts

**Learning goals:** Assemble the conceptual component groups used by XBeach and connect them to focused tutorials.

**Prerequisites:** Complete Journey 5 and have `rompy-xbeach` installed.

**Execution contract:** This lesson constructs validated component objects but does not generate or run an XBeach workspace.

```python
from rompy_xbeach.components.physics import Physics
from rompy_xbeach.components.physics.wavemodel import Surfbeat
from rompy_xbeach.components.sediment import Sediment
from rompy_xbeach.components.output import Output

physics = Physics(wavemodel=Surfbeat())
sediment = Sediment()
output = Output()
assert physics.wavemodel.model_type == 'surfbeat'
print('Validated component groups:', type(physics).__name__, type(sediment).__name__, type(output).__name__)
```
