# Journey 3 — Configure XBeach declaratively with YAML

**Learning goals:** Represent a small XBeach run as serialisable YAML and understand what configuration remains separate from execution.

**Prerequisites:** Complete Journey 2 and know basic YAML mappings.

**Execution contract:** This lesson validates a configuration document only; it does not load external data or run XBeach.

```python
import yaml

configuration = {
    'run_id': 'xbeach_learning',
    'period': {'start': '2023-01-01T00:00', 'end': '2023-01-01T06:00', 'interval': '1h'},
    'config': {'model_type': 'xbeach', 'grid': {'model_type': 'regular'}},
}
serialized = yaml.safe_dump(configuration, sort_keys=False)
assert yaml.safe_load(serialized) == configuration
print(serialized)
```
