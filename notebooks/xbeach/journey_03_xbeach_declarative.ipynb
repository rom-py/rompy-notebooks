# Journey 3 — Configure XBeach declaratively with YAML

**Learning goals:** Represent a small XBeach run as serialisable YAML and understand what configuration remains separate from execution.

**Prerequisites:** Complete Journey 2 and know basic YAML mappings.

**Execution contract:** This lesson validates a configuration document only; it does not load external data or run XBeach.

## Why this matters: portable configuration

**Without Rompy:** sharing an XBeach experiment may require readers to reconstruct a Python object graph or copy setup scripts.

**With Rompy:** a declarative file records the run contract in a form that can be reviewed, versioned, and shared, while data access and optional execution remain separate.


````python
import yaml

configuration = {
    'run_id': 'xbeach_learning',
    'period': {'start': '2023-01-01T00:00', 'end': '2023-01-01T06:00', 'interval': '1h'},
    'config': {'model_type': 'xbeach', 'grid': {'model_type': 'regular'}},
}
serialized = yaml.safe_dump(configuration, sort_keys=False)
assert yaml.safe_load(serialized) == configuration
print(serialized)
````

## 2. Use the YAML from the Rompy CLI

The complete YAML shown in the [declarative example](../example-declarative/) can also be used without opening a Python environment. From the directory containing the file:

```bash
rompy validate example-declarative.yml
rompy generate example-declarative.yml
```

`generate` prepares the XBeach workspace. To validate, generate, and launch XBeach in one command, provide a backend configuration for the environment containing the XBeach executable:

```bash
rompy run example-declarative.yml --backend-config backend.yml
```

This is the same validated configuration demonstrated above, with the CLI providing the execution entry point.
