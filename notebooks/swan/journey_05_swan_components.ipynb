# Journey 5 — Configure SWAN components and outputs

**Learning goals:** Assemble representative startup, physics, numerics, boundary, and output components into a `SwanConfig`.

**Prerequisites:** [Journey 4](../journey_04_swan_data/).


## Why this matters: composable model controls

**Without Rompy:** SWAN startup, physics, numerics, boundary, and output choices can drift across hand-written files and scripts.

**With Rompy:** typed components keep those separate concerns inspectable and validated as one coherent configuration before a workspace is generated.


## 1. Startup and physics

SWAN components render model-native commands. Composition keeps each concern inspectable and validated.


```python
from rompy_swan.components.group import STARTUP, PHYSICS
from rompy_swan.components.startup import PROJECT, SET, MODE, COORDINATES
from rompy_swan.subcomponents.startup import SPHERICAL
from rompy_swan.components.physics import GEN3, BREAKING_CONSTANT
from rompy_swan.subcomponents.physics import WESTHUYSEN

startup = STARTUP(
    project=PROJECT(name="Rompy SWAN", nr="run1"),
    set=SET(level=0.0, depmin=0.05, direction_convention="nautical"),
    mode=MODE(kind="nonstationary", dim="twodimensional"),
    coordinates=COORDINATES(kind=SPHERICAL()),
)
physics = PHYSICS(
    gen=GEN3(source_terms=WESTHUYSEN()),
    breaking=BREAKING_CONSTANT(alpha=1.0, gamma=0.73),
)
print(startup.render())
print(physics.render())
```

## 2. Numerics and boundary conditions


```python
from rompy_swan.components.numerics import PROP
from rompy_swan.subcomponents.numerics import BSBT
from rompy_swan.components.boundary import INITIAL
from rompy_swan.subcomponents.boundary import DEFAULT

propagation = PROP(scheme=BSBT())
initial = INITIAL(kind=DEFAULT())
print(propagation.render())
print(initial.render())
```

## 3. Describe outputs

Output components define which fields, point time series, and spectra SWAN should write. This is separate from post-processing: output configuration tells SWAN what to produce, while later analysis reads those files.


```python
from rompy_swan.components.output import POINTS, QUANTITY, QUANTITIES, BLOCK

points = POINTS(sname="stations", xp=[115.5], yp=[-31.5])
quantities = QUANTITIES(quantities=[QUANTITY(output=["hsign", "tps"])])
block = BLOCK(
    sname="COMPGRID",
    fname="outgrid.nc",
    output=["depth", "wind", "hsign"],
)
print(points.render())
print(quantities.render())
print(block.render())
```
